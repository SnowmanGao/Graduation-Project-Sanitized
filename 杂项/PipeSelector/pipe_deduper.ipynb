```python
from collections import defaultdict


def parser(code: str) -> list[str]:
    return code.split("-")


def check_dup(codes: list[str]) -> bool:
    seen = set()
    dup = defaultdict(list)

    for code in codes:
        parts = parser(code)
        if len(parts) < 2:
            raise ValueError(f"管道编号格式错误：{code}")
        material = parts[0]
        number = parts[1]
        key = (material, number)
        if key in seen:
            dup[key].append(code)
        else:
            seen.add(key)

    if dup:
        print("存在重复编号：")
        for (material, number), repeated_codes in dup.items():
            print(f"{material}-{number}")
            for code in repeated_codes:
                print(f"  {code}")
        return True

    print("不存在重复编号")
    return False


check_dup(inp)
```
